```python
import pandas as pd  # Library for data manipulation and analysis
import plotly.express as px  # Library for interactive visualizations
from prettytable import PrettyTable  # Library to display tabular data nicely
from IPython.display import display, HTML  # To enhance display options in Jupyter Notebook

 
# Read the dataset into a pandas DataFrame
data = pd.read_csv('ad_click_data.csv')
 
# ==============================================================================
# 1. NUMERICAL FEATURE SUMMARY & DISTRIBUTION
# ==============================================================================
 
# Define numeric columns for detailed statistical analysis and visualization
numeric_columns = [
    'Age', 'Time_Spent_on_Site', 'Time_of_Day', 'Previous_Clicks'
]
 
# Initialize PrettyTable to display numeric summaries clearly
summary_table = PrettyTable()
 
# Define column headers for the PrettyTable
summary_table.field_names = ["Feature", "Mean", "Median", "Std Dev", "Min", "Max", "Missing %"]
 
# Loop through each numeric column to calculate summary statistics
for col in numeric_columns:
    mean = round(data[col].mean(), 2)  # Average value of the feature
    median = round(data[col].median(), 2)  # Middle value (50th percentile)
    std = round(data[col].std(), 2)  # Standard deviation (measure of variability)
    min_val = round(data[col].min(), 2)  # Minimum observed value
    max_val = round(data[col].max(), 2)  # Maximum observed value
    missing_percent = round(data[col].isnull().mean() * 100, 2)  # Percentage of missing data
 
    # Add the calculated statistics for the current feature to the PrettyTable
    summary_table.add_row([col, mean, median, std, min_val, max_val, missing_percent])
 
# Display the PrettyTable with summary statistics
print("Numerical Feature Summary:")
print(summary_table)
 
# Generate interactive histograms using Plotly for each numeric feature
print("\nGenerating Histograms for Numerical Features...")
for col in numeric_columns:
    # Create histogram with boxplot marginal to visualize distribution and detect outliers
    fig = px.histogram(
        data,  # Data source
        x=col,  # Feature to plot
        title=f'Distribution of {col}',  # Title of the histogram
        marginal="box",  # Display boxplot on margin to highlight distribution statistics
        opacity=0.75,  # Transparency of histogram bars
        nbins=50,  # Number of bins in the histogram
        color_discrete_sequence=['teal']  # Color of bars in histogram
    )
 
    # Adjust layout to improve visualization
    fig.update_layout(bargap=0.1)  # Adjust gap between bars for better readability
 
    # Display the interactive histogram
    fig.show()
    
# ==============================================================================
# 2. CATEGORICAL FEATURE SUMMARY & VISUALIZATION
# ==============================================================================
 
# Define categorical columns to analyze
categorical_columns = [
    'Gender', 'Geographic_Location', 'Ad_Position', 'Device_Type', 
    'Ad_Category', 'Day_of_Week'
]
 
# Display value counts for each categorical feature
print("\nCategorical Feature Value Counts:")
for col in categorical_columns:
    print(f"\n--- Value Counts for '{col}' ---")
    print(data[col].value_counts())
    print("-" * 30)
 
# Generate bar plots to show click-through rate for each category
print("\nGenerating Click-Through Rate Bar Plots for Categorical Features...")
for col in categorical_columns:
    # Group by the categorical feature and calculate the mean of 'Clicked' (which is the CTR)
    ct_data = data.groupby(col)['Clicked'].mean().reset_index()
    ct_data['CTR (%)'] = ct_data['Clicked'] * 100 # Convert to percentage
 
    # Create a bar chart showing the CTR for each category
    fig = px.bar(
        ct_data,
        x=col,
        y='CTR (%)',
        title=f'Click-Through Rate (CTR) by {col}',
        color_discrete_sequence=['teal'],
        text='CTR (%)' # Display the CTR value on top of the bars
    )
 
    # Customize the layout
    fig.update_traces(texttemplate='%{text:.2f}%', textposition='outside')
    fig.update_layout(uniformtext_minsize=8, uniformtext_mode='hide')
 
    fig.show()
 
# ==============================================================================
# 3. TARGET VARIABLE DISTRIBUTION
# ==============================================================================
 
# Calculate and display the distribution of the target variable 'Clicked'
clicked_counts = data['Clicked'].value_counts(normalize=True) * 100
print("\n--- 'Clicked' Target Variable Distribution ---")
print(f"Percentage of non-clicks (0): {clicked_counts[0]:.2f}%")
print(f"Percentage of clicks (1): {clicked_counts[1]:.2f}%")
 
# Create a pie chart to visualize the class imbalance
fig = px.pie(
    names=clicked_counts.index,
    values=clicked_counts.values,
    title='Distribution of the Target Variable (Clicked)',
    labels={0: 'Not Clicked', 1: 'Clicked'},
    hole=0.4,
    color_discrete_sequence=['lightcoral', 'darkgreen']
)
fig.show()
 
print("\nEDA script complete. Check the generated plots and console output for insights.")
```

## **Missing Value Treatments: Importance, Strategies, and Considerations**<br>
 
**Why Missing Value Treatment is Important:**<br>
Missing data is a common issue in online advertising datasets, often resulting from user privacy settings, ad blocker software, or logging errors. Handling these missing values is crucial because:<br>
 
- **Flawed Business Decisions:** Unaddressed missing data can lead to inaccurate model predictions, resulting in wasted ad spend and reduced revenue.<br>
- **Data Integrity and Model Accuracy:** Proper treatment ensures that the predictive model is trained on a complete and reliable dataset, leading to more accurate insights and improved CTR optimization.<br>
 
---
 
**Types of Missing Data and Their Implications:**<br>
 
- **Missing Completely at Random (MCAR):**<br>
  - Example: A random server glitch causes some 'Time_Spent_on_Site' records to be lost, unrelated to user behavior.<br>
  - Simplest type to handle.<br>
 
- **Missing at Random (MAR):**<br>
  - Example: A user's 'Age' is more likely to be missing if their Device_Type is 'Mobile'.<br>
  - Missingness depends on an **observed variable**, not the missing value itself.<br>
 
- **Missing Not at Random (MNAR):**<br>
  - Example: Older users may be less likely to provide their age due to privacy concerns.<br>
  - Missingness depends on the **value of the missing data itself**, making it most complex.<br>
 
---
 
**How to Detect the Type of Missingness:**<br>
 
- **MCAR:**<br>
  - Use statistical tests (e.g., chi-square) to check independence between missingness and other features.<br>
 
- **MAR:**<br>
  - Create missingness indicators (e.g., `is_age_missing`) and test correlation with other variables like Device_Type.<br>
 
- **MNAR:**<br>
  - Harder to detect programmatically.<br>
  - Requires **business intuition** or prior research about user behavior.<br>
 
---
 
**Common Strategies for Missing Value Treatment:**<br>
 
- **Simple Imputation:**<br>
  - **Median Imputation:** Replace missing Age, Time_Spent_on_Site, or Previous_Clicks with the median (robust for skewed data).<br>
  - **Mode Imputation:** For categorical data (not applicable here).<br>
 
- **Advanced Imputation:**<br>
  - **Regression-Based Imputation:** Predict missing values using related features (e.g., Device_Type, Ad_Position).<br>
  - **KNN Imputation:** Use nearest neighbors to impute missing values, leveraging dataset structure.<br>
 
---
 
**Potential for Injecting Bias:**<br>
 
- Wrong imputation methods (e.g., mean for skewed Previous_Clicks) can distort distributions and bias model coefficients.<br>
- Always check post-imputation distributions to ensure realism.<br>
 
---
 
**Recommended Best Practices for This Project:**<br>
 
- **Start Simple:** Use **median imputation** for Age, Time_Spent_on_Site, and Previous_Clicks.<br>
- **Validate:** Perform EDA after imputation to ensure distributions remain stable.<br>
- **Benchmark:** Build a second model with advanced imputation (e.g., KNN) to compare performance.<br>
 
By carefully identifying missingness types and applying the right imputation strategies, we can ensure our CTR prediction model remains accurate, robust, and business-relevant.<br>

```python
data_simple_imputed = data.copy()
```

```python
data_simple_imputed["Age"] = data_simple_imputed["Age"].fillna(data_simple_imputed["Age"].median())

data_simple_imputed["Time_Spent_on_Site"] = data_simple_imputed["Time_Spent_on_Site"].fillna(data_simple_imputed["Time_Spent_on_Site"].median())

data_simple_imputed["Previous_Clicks"] = data_simple_imputed["Previous_Clicks"].fillna(data_simple_imputed["Previous_Clicks"].median())
```

```python
# Import necessary libraries
import pandas as pd
from scipy.stats import chi2_contingency
 
# Assuming 'data_simple_imputed' is your preprocessed DataFrame
# Create a contingency table (crosstab)
# This table shows the frequency of 'Clicked' for each 'Device_Type'
contingency_table = pd.crosstab(data_simple_imputed['Device_Type'], data_simple_imputed['Clicked'])
 
# Perform the Chi-Square test
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)
 
# Display the results
print("Chi-Square Test of Independence for Device_Type and Clicked:")
print("---------------------------------------------------------")
print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print("\nContingency Table:")
print(contingency_table)
 
# Business-oriented conclusion
if p_value < 0.05:
    print("\nConclusion: Statistically significant association found. The type of device a user is on is strongly related to whether they click on an ad.")
else:
    print("\nConclusion: No statistically significant association. Device type does not appear to be related to ad clicks.")
```

## **Business Interpretation of Chi-Square Test: Device_Type vs. Clicked**
 
The results of the Chi-Square test, with a p-value of 0.0000, provide strong statistical evidence that there is a significant relationship between a user's Device_Type and their likelihood of clicking on an ad. This confirms our initial EDA findings.
 
- Business Impact: The contingency table is a crucial tool for AdVantage Analytics. It shows that mobile users are the most valuable audience, as every ad served to a mobile device resulted in a click. In contrast, desktop and tablet users clicked on ads far less frequently. This insight should guide AdVantage to prioritize mobile-first ad campaigns and optimize their ad-serving algorithms to target mobile devices, leading to a direct increase in overall Click-Through Rate (CTR) and client ROI.

```python
# Import necessary libraries
import pandas as pd
from scipy.stats import mannwhitneyu
 
# Assuming 'data_simple_imputed' is your preprocessed DataFrame
# Separate the 'Previous_Clicks' data into two groups based on the 'Clicked' target variable.
previous_clicks_clicked = data_simple_imputed[data_simple_imputed['Clicked'] == 1]['Previous_Clicks'].dropna()
previous_clicks_not_clicked = data_simple_imputed[data_simple_imputed['Clicked'] == 0]['Previous_Clicks'].dropna()
 
# Perform the Mann-Whitney U test
u_stat, p_value = mannwhitneyu(previous_clicks_clicked, previous_clicks_not_clicked, alternative='two-sided')
 
# Display the results
print("Mann-Whitney U Test for Previous_Clicks:")
print("------------------------------------------")
print(f"U-statistic: {u_stat:.4f}")
print(f"P-value: {p_value:.4f}")
 
# Business-oriented conclusion
if p_value < 0.05:
    print("\nConclusion: Statistically significant difference found. The distribution of previous clicks is significantly different for users who clicked versus those who did not.")
else:
    print("\nConclusion: No statistically significant difference. The number of previous clicks does not significantly differ between users who clicked and those who did not.")
```

## **Business Interpretation of Mann-Whitney U Test: Previous_Clicks vs. Clicked**
 
The Mann-Whitney U test returned a p-value of 0.3818. Since this is greater than the standard significance level of 0.05, we cannot conclude that a user's history of clicking on ads is a statistically significant predictor of their current click behavior.
 
- Business Impact: This finding suggests that for this specific dataset, a user's past click behavior is not a reliable indicator for predicting future clicks. AdVantage should not heavily weigh the Previous_Clicks feature in their final model, as it lacks a strong statistical relationship with the target variable. Instead, the business should pivot its focus to the more reliable and impactful insights from the Device_Type analysis, which has proven to be a primary driver of clicks.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
 
# Load the ORIGINAL data, including its missing values
data_model = pd.read_csv('ad_click_data.csv')
 
# Define the predictors available at prediction time
numeric_features = [
    'Age',
    'Time_of_Day',
    'Time_Spent_on_Site',
    'Previous_Clicks'
]
 
categorical_features = [
    'Gender',
    'Geographic_Location',
    'Ad_Position',
    'Device_Type',
    'Ad_Category',
    'Day_of_Week'
]
 
features = numeric_features + categorical_features
target = 'Clicked'
 
X = data_model[features]
y = data_model[target]
 
# Split BEFORE learning imputation values or encoding categories
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
 
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
 
print("\nTraining target distribution (%):")
print((y_train.value_counts(normalize=True) * 100).round(2))
 
print("\nTesting target distribution (%):")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
 
train_percent = (
    y_train.value_counts(normalize=True).sort_index() * 100
)
test_percent = (
    y_test.value_counts(normalize=True).sort_index() * 100
)
 
comparison = pd.DataFrame({
    'Training Set (%)': train_percent,
    'Testing Set (%)': test_percent
}).rename(index={0: 'Not Clicked', 1: 'Clicked'})
 
print(comparison.round(2))
 
comparison.plot(kind='bar', figsize=(8, 5))
plt.title('Target Class Distribution: Train vs Test')
plt.xlabel('Target Class')
plt.ylabel('Percentage of Observations')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
 
# Numeric columns: learn medians from training data only
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median'))
])
 
# Categorical columns: learn frequent values and categories
# from training data only
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
 
# Apply the appropriate preprocessing to each column group
preprocessor = ColumnTransformer(transformers=[
    ('numeric', numeric_transformer, numeric_features),
    ('categorical', categorical_transformer, categorical_features)
])
 
# Combine preprocessing and Logistic Regression into one pipeline
log_reg_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        solver='liblinear',
        random_state=42,
        class_weight='balanced',    #class_weight='balanced' makes sure that the model is not biased towards one class and dishes out a higher penalty for misclassifying the minority class.
        max_iter=1000
    ))
])
 
print("Preprocessing and Logistic Regression pipeline created.")
```

```python
# Import the necessary library for Logistic Regression
from sklearn.linear_model import LogisticRegression
 
# Initialize the Logistic Regression model
# We'll use the 'liblinear' solver, which is good for small datasets
# and handles L1 and L2 regularization well.
# The 'C' parameter is the inverse of regularization strength.
# We'll also use class_weight='balanced' to handle our class imbalance.
# log_reg_model = LogisticRegression(solver='liblinear', random_state=42, class_weight='balanced')    do not run this again, otherwise it will reinitialise the model object and forget the previous training. 
 
# Train the model on the preprocessed training data
# X_train_final contains all the numerical and one-hot encoded features.
# y_train is our target variable.
log_reg_model.fit(X_train, y_train)
preprocessor = log_reg_model.named_steps['preprocessor']
 
feature_names = preprocessor.get_feature_names_out()
 
print("Original input features:", X_train.shape[1])
print("Features after preprocessing:", len(feature_names))
 
# The model has now been trained.
# You can view the coefficients and intercept to understand the model's parameters.
print("Model training complete.")
print("\nModel Coefficients:")
log_reg_model.named_steps['classifier'].coef_
print("\nModel Intercept:")
log_reg_model.named_steps['classifier'].intercept_
```

this will increase precision of class 1 but may also miss actual clickers.(reduces recall).
